## 3. Document DSM-5 label drift and route ambiguity (Rise 8.3)

The two legacy example fields, `opioid_abuse_history` and `opioid_dependence_history`, **cannot each be renamed into a verified DSM-5 diagnosis** without reviewing the source chart and diagnostic criteria. Preserve source field names and definitions; propose a *candidate* current construct, leave any unsupported concept ID blank, and route ambiguous cases to a qualified human reviewer. Foundational completes both rows. Applied additionally edits the demonstration routing function; it never adjudicates real patients here.

```python
legacy_mapping=pd.DataFrame([
 {"source_field":"opioid_abuse_history","source_system":"legacy DSM-IV chart review","source_definition":"[Check chart-review rule]","candidate_current_construct":"[Candidate DSM-5 concept, if supported]","verified_id":"","decision":"needs human review","mapping_rule":"[Explain when evidence supports a mapping, and what remains ambiguous]","reviewer":"[Clinical reviewer role]","source_reference":"[Codebook/chart rule or scenario]","version":"draft-01"},
 {"source_field":"opioid_dependence_history","source_system":"legacy DSM-IV chart review","source_definition":"[Check chart-review rule]","candidate_current_construct":"[Candidate DSM-5 concept, if supported]","verified_id":"","decision":"needs human review","mapping_rule":"[Explain when evidence supports a mapping, and what remains ambiguous]","reviewer":"[Clinical reviewer role]","source_reference":"[Codebook/chart rule or scenario]","version":"draft-01"},
])
display(legacy_mapping.T)

def route_legacy_mapping(abuse_history,dependence_history,documentation_available=False):
    """Teaching demonstration: ambiguous positive legacy labels require human review."""
    if abuse_history not in (0,1) or dependence_history not in (0,1):
        return "needs review: missing or invalid legacy value"
    if abuse_history or dependence_history:
        return "needs review: verify current diagnostic criteria and chart context"
    # Two negative legacy flags cannot prove no DSM-5 disorder.
    return "not documented by these legacy fields; do not infer absence"

example_rows=pd.DataFrame({"legacy_abuse":[1,0,0,None],"legacy_dependence":[0,1,0,0]})
example_rows["routing_result"]=[route_legacy_mapping(a,b) for a,b in zip(example_rows.legacy_abuse,example_rows.legacy_dependence)]
display(example_rows)
if PATHWAY=="Applied": print("APPLIED: implement your own documented, validated rules and reviewer handoff. This conservative demonstration does not assign diagnoses.")
```
