## What is a Notebook?

A notebook is a workspace made of **cells**. Each cell holds one small part of your work.

There are two main kinds of cells in this notebook:

- **Text cells** hold explanations, instructions, questions, and reflections. You can create a new text cell by clicking the `+ Text` button.
- **Code cells** hold Python code that you can run. You can create a new code cell by clicking the `+ Code` button.
  - Code cells can also include **comments**, which are notes that explain the code for the reader. **Comments are not run as code**. In Python, a line that starts with `#` is a comment.

When you run a code cell, Colab shows the **output** underneath. Output is what the notebook gives back to you as a result of the code you ran. Sometimes the output is what you hoped for, such as a message, a number, or a table. Sometimes the output is an **error**.

An **error** is a message that tells you something in the code needs attention. Errors are normal in coding and are a part of learning.

Notebooks are meant to be interactive. You can run one step, look at the result, change something, and run it again. In notebooks, code cells often **build on earlier cells**. This means later cells may depend on something you already ran above. If you change an earlier cell and rerun it, the result in later cells may change too.

## Important Concepts: Variables and Exact Names

A **variable** is a name that stores something so you can use it again later.

For example, in this notebook, `df` is the variable that stores our table. When later code uses `df`, it is referring back to that same table.

Another helpful beginner idea is that in coding, **names usually need to match exactly**.

This applies to many parts of coding, including:
- variable names
- column names
- function names
- library names

For example, if your table is named `df`, later code must also use `df`, not `DF` or `dataframe`. If a column is named `naloxone_kits_distributed`, your code needs to use that exact name as well.

Even small differences, such as:
- capital letters
- missing underscores
- extra spaces
- misspellings

can cause code to break or give an error.

A helpful habit is to copy names directly from earlier code when you can, instead of typing them again from memory.

You do **not** need to memorize these ideas right away. They are here to help the notebook make more sense as you work through it.

## Step 3A: Create a Small Fictional Dataset

## Step 3C: Use Python to Answer a Simple Public Health Question

Run the code cell below to answer a simple question:  
**Which county has the highest number of naloxone kits distributed?**

You do not need to memorize this code. Focus on the relationship between the **question**, the **code**, and the **result**.

```python
# Sort the table from highest to lowest by naloxone kit distribution.
highest_row = df.sort_values("naloxone_kits_distributed", ascending=False).iloc[0]

# Print the county from the top row after sorting.
print("County with highest naloxone kit distribution:", highest_row["county"])

# Print the number of kits distributed for that county.
print("Number of kits distributed:", highest_row["naloxone_kits_distributed"])
```

## AI-Assisted Prompting Guidelines

Gemini can be a useful learning partner in Colab when you use it with **clear prompts** and **careful checking**.

### Prompting Guidelines for Coding Help

When you ask Gemini for coding help, stronger prompts usually do these things:

1. **State the task clearly**  
   Say exactly what you want the code to do.

2. **Give the right context**  
   Name the dataframe, variables, or columns Gemini should use.

3. **Use exact names from the notebook**  
   Tell Gemini to use the exact same variable names already shown in your code.

4. **Add helpful constraints**  
   For example, say you want beginner-friendly Python, direct output, or a simple explanation.

5. **Say what a good result should look like**  
   Ask Gemini to print the answer clearly and tell you what output to expect.

6. **Ask how to verify the result**  
   Do not stop at getting code. Ask how to check whether the answer is correct.

7. **Break bigger tasks into smaller steps**  
   If the task feels big, first ask for ideas, then choose one, then ask for code.

8. **Revise the prompt if needed**  
   If the first answer is too advanced, unclear, or off-task, ask Gemini to simplify, correct, or revise it.

   
Summary: A helpful coding prompt usually includes these parts:

1. **The task** — what you want the code to do  
2. **The context** — what data, variable, or columns Gemini should use  
3. **Your constraints** — for example, that you want beginner-friendly code or a direct answer in Colab  
4. **Your check** — what result you expect, or how you will verify that it worked

## Step 4A: Use Gemini to Generate Code in Colab

In this first practice example, the goal is **not** to let AI do everything. The goal is to practice writing a clear prompt, running the code, and checking whether the result really matches your question.

### Prompt to Try in Colab's Built-In AI (Gemini)

Open Gemini in Colab by clicking the blue circle with a four pointed star at the bottom of the notebook, and try prompting it.

A **weak prompt** would be something like:

> Find the highest value in the data.

A **strong prompt** would be something like this:

> Using the existing dataframe `df` in this notebook, write beginner-friendly Python code that finds the county with the highest value in the `naloxone_kits_distributed` column. Use the exact dataframe and column names shown in the notebook. Print the county name and the number of kits distributed. Then explain what output I should expect and how I can verify that the result is correct.


### Components of the Stronger Prompt

A stronger prompt works better because it:
- says **what task** to do
- says **what data** to use
- tells Gemini to use **the exact names from the notebook**
- asks for **beginner-friendly code**
- asks **what output to expect**
- asks **how to verify** the result

### Verify the Result Carefully

After Gemini gives you code and you run it, check these specific things:

1. **Dataframe check:** Did the code use `df`, not a different variable name?  
2. **Column-name check:** Did it use `naloxone_kits_distributed` exactly, with the same spelling and underscores?  
3. **Question check:** Did it find the county with the **highest** value in that column, not a different summary such as an average or total?  
4. **Table check:** Based on the table above, did it return **county C** and **150**?  
5. **Output check:** Did the printed result match what Gemini said you should expect?  
6. **Explanation check:** When Gemini explains the code, does that explanation match what the code actually did?

If something is off, ask a follow-up prompt such as:

> Please revise the code so it uses the existing dataframe `df` and the exact column name `naloxone_kits_distributed`. Keep the code beginner-friendly, print the result clearly, and explain what changed.

## Other Ways Gemini Can Help You While You Are Learning

Gemini can be a useful learning partner in Colab when you use it thoughtfully. Here are some beginner-friendly prompt ideas:

- **Explain code in plain language**  
  *"Explain what this code does in beginner-friendly language. Please go line by line."*

- **Simplify code**  
  *"Rewrite this in simpler Python that a beginner can follow. Keep the same result."*

- **Work through an error**  
  *"I got this error after running the code below. Explain what it means, identify the likely cause, and help me fix it without changing the goal of the analysis."*

- **Suggest a next step**  
  *"Suggest 2 or 3 beginner-friendly analyses I could try with `df`. For each one, briefly explain what I would learn."*

### Important Reminder

Gemini can help you learn, but it can also make mistakes. Always check:
- whether the code matches your question
- whether the names match what is in your notebook
- whether the result makes sense

## Step 4B: Explore Data by Conversing with Gemini

Now you will use Gemini to suggest additional beginner-friendly ways to explore the dataset.

### Step 1: Ask for a Few Small Ideas

2. **Task check:** Did the code answer the analysis you chose, rather than switching to a different question?  
3. **Output check:** Did the output match what Gemini said you should expect?  
4. **Table check:** Can you connect the result back to the values in the table you can see?  
5. **Explanation check:** Does Gemini's explanation match what the code actually did?

### Step 4: Ask Gemini to Explain the Result

After that, ask a follow-up such as:

> Explain what this code did, why this analysis is useful for this dataset, and what the result means in beginner-friendly language.
